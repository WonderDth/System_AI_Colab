# Systemy AI – Laboratorium 4: pingwiny

Pipeline dla zbioru penguins, na wzór notatnika z Titanikiem. Przewidujemy gatunek (species).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report
```

## 1. Wczytanie danych

```python
df = sns.load_dataset("penguins")
print(df.shape)
df.head()
```

```python
df.info()
```

## 2. EDA

```python
print(df["species"].value_counts(normalize=True).round(3))
df.isna().sum()
```

```python
num_features = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
print(df[num_features].describe().round(1))
print("skośność:", df[num_features].skew().round(2).to_dict())
```

```python
print(df.groupby("species")[num_features].mean().round(1))
print(pd.crosstab(df["island"], df["species"]))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.scatterplot(data=df, x="bill_length_mm", y="bill_depth_mm", hue="species", ax=axes[0])
axes[0].set_title("Długość i wysokość dzioba")
sns.histplot(data=df, x="flipper_length_mm", hue="species", bins=25, ax=axes[1])
axes[1].set_title("Długość płetwy")
fig.tight_layout()
plt.show()
```

Wnioski z EDA:
- najwięcej jest Adelie (ok. 44%), najmniej Chinstrap (ok. 20%), więc przy podziale używam stratify i porównuję wynik z baseline,
- 2 pingwiny nie mają żadnych pomiarów, a w sex brakuje 11 wartości,
- island i sex są tekstowe, więc trzeba one-hot,
- masa jest w tysiącach gramów, a dziób ma kilkanaście–kilkadziesiąt mm, więc potrzebna standaryzacja,
- Gentoo ma dłuższą płetwę i większą masę, a Adelie od Chinstrap odróżnia głównie długość dzioba,
- Gentoo jest tylko na Biscoe, Chinstrap tylko na Dream.

```python
puste = df[num_features].isna().all(axis=1)
print(df[puste])
df = df[~puste].reset_index(drop=True)
print(df.shape)
```

Usuwam 2 wiersze bez pomiarów, bo po imputacji byłyby to wymyślone pingwiny z prawdziwą etykietą. Wiersze z brakiem samej płci zostawiam, bo mają wszystkie pomiary. Kolumn nie usuwam, bo nie ma tu duplikatów ani kolumny, która zdradza etykietę.

## 3. Podział train / test

```python
cat_features = ["island", "sex"]
features = num_features + cat_features
target = "species"

X_train, X_test, y_train, y_test = train_test_split(
    df[features], df[target],
    test_size=0.25,
    stratify=df[target],
    random_state=42,
)
print(X_train.shape, X_test.shape)
print(pd.DataFrame({"train": y_train.value_counts(normalize=True),
                    "test": y_test.value_counts(normalize=True)}).round(3))
```

## 4. Przygotowanie danych

```python
przygotowanie = ColumnTransformer([
    ("liczby", make_pipeline(SimpleImputer(strategy="median"), StandardScaler()), num_features),
    ("kategorie", make_pipeline(SimpleImputer(strategy="most_frequent"),
                                OneHotEncoder(handle_unknown="ignore")), cat_features),
])

model = make_pipeline(przygotowanie, LogisticRegression(max_iter=1000))
model
```

## 5. Trenowanie

```python
model.fit(X_train, y_train)

lr = model.named_steps["logisticregression"]
nazwy = model.named_steps["columntransformer"].get_feature_names_out()
wagi = pd.DataFrame(lr.coef_, index=lr.classes_, columns=nazwy).T
wagi.round(2)
```

Są 3 klasy, więc model ma osobny zestaw wag dla każdego gatunku. Duża dodatnia waga przy bill_length_mm jest u Chinstrap, a ujemna u Adelie, co zgadza się z EDA.

## 6. Predykcja

```python
y_pred = model.predict(X_test)
proba = pd.DataFrame(model.predict_proba(X_test), columns=model.classes_, index=X_test.index)

podglad = X_test.head(5).join(proba.head(5).round(2))
podglad["predykcja"] = y_pred[:5]
podglad["prawda"] = y_test.head(5).values
podglad
```

## 7. Ewaluacja

```python
baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)

def regula(X):
    return np.where(X["flipper_length_mm"] >= 207, "Gentoo",
           np.where(X["bill_length_mm"] >= 44, "Chinstrap", "Adelie"))

wyniki = pd.Series({
    "baseline (zawsze Adelie)": baseline.score(X_test, y_test),
    "reguła z EDA": (regula(X_test) == y_test).mean(),
    "pipeline – train": model.score(X_train, y_train),
    "pipeline – test": model.score(X_test, y_test),
})
wyniki.round(3)
```

```python
fig, ax = plt.subplots(figsize=(5, 4.5))
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred, labels=model.classes_),
                       display_labels=model.classes_).plot(ax=ax, colorbar=False)
ax.set_title("Macierz pomyłek (test)")
plt.show()

print(classification_report(y_test, y_pred, digits=3))
```

Baseline daje 0.44, czyli tyle, ile jest Adelie. Prosta reguła z płetwy i dzioba (progi odczytane z wykresów) pomyliła się tylko raz (0.988), a pipeline trafił wszystkie 86 pingwinów (1.000). To różnica jednego pingwina, więc na tych danych model niewiele dodaje do prostej reguły. Wysoki wynik wynika z tego, że gatunki bardzo się różnią wymiarami, a test jest mały.

## 8. Iteracja

```python
def zbuduj(num, cat, skalowanie=True):
    kroki = [SimpleImputer(strategy="median")]
    if skalowanie:
        kroki.append(StandardScaler())
    czesci = []
    if num:
        czesci.append(("liczby", make_pipeline(*kroki), num))
    if cat:
        czesci.append(("kategorie", make_pipeline(SimpleImputer(strategy="most_frequent"),
                                                  OneHotEncoder(handle_unknown="ignore")), cat))
    return make_pipeline(ColumnTransformer(czesci), LogisticRegression(max_iter=1000))


bez_dzioba = ["bill_depth_mm", "flipper_length_mm", "body_mass_g"]

warianty = {
    "pełny": zbuduj(num_features, cat_features),
    "bez długości dzioba": zbuduj(bez_dzioba, cat_features),
    "tylko wyspa + płeć": zbuduj([], cat_features),
    "bez skalowania": zbuduj(num_features, cat_features, skalowanie=False),
}
pd.Series({nazwa: m.fit(X_train, y_train).score(X_test, y_test) for nazwa, m in warianty.items()}).round(3)
```

```python
m = warianty["bez długości dzioba"]
pd.crosstab(y_test, m.predict(X_test), rownames=["prawda"], colnames=["predykcja"])
```

Bez długości dzioba model pomylił się 9 razy (7 Chinstrap uznał za Adelie, 2 Adelie za Chinstrap), a Gentoo rozpoznał bezbłędnie. Wszystkie pomyłki są między Adelie i Chinstrap, bo to właśnie długość dzioba je rozróżnia. Sama wyspa i płeć to za mało, bo na Dream są Adelie i Chinstrap, a na Biscoe Adelie i Gentoo. Bez skalowania wynik jest taki sam (1.000), ale pojawia się ConvergenceWarning - algorytm nie zdążył dojść do rozwiązania, bo cechy mają bardzo różne skale.

Wybieranie wariantu po wyniku na teście nie jest do końca poprawne, do tego powinien być osobny zbiór walidacyjny albo walidacja krzyżowa.
